```python
print(df.describe())
```

```python
range_by_make = df.groupby("Make")["Electric Range"].mean().sort_values(ascending=False)
print(range_by_make)
```

```python
Porsche_car = df[df["Make"]=="PORSCHE"]
print(Porsche_car.head())
```

```python
make_avg = df.groupby("Make")["Base MSRP"].mean().reset_index()
df = pd.merge(df,make_avg,on= "Make",suffixes=('','_Make_Avg'))
print(df.head())
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Electric Range"],bins=20,color="skyblue",edgecolor="black")
plt.title("Distribution of Electric vehicle Range")
plt.xlabel("Range(miles)")
plt.ylabel("Make")
plt.show()
```

```python
top_makes = range_by_make.head(10)
plt.figure(figsize=(10,6))
plt.bar(top_makes.index,top_makes.values,color = "green")
plt.xticks(rotation = 45)
plt.title("Average Electric Range by Make")
plt.xlabel("Make")
plt.ylabel("Average Range(miles)")
plt.show()
```

```python
plt.figure(figsize=(6,6))
df["Electric Vehicle Type"].value_counts().plot.pie(autopct = "%1.1f%%")
plt.title("Distribution of Electric Vehicle Types")
plt.show()
```

```python
plt.figure(figsize = (10,6))
sns.boxplot(x="Make",y = "Base MSRP",data=df)
plt.xticks(rotation =90)
plt.title("MSRP Distribution by Make")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df[['Base MSRP','Electric Range']].corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[["Base MSRP","Electric Range"]])
plt.show()
```

```python
sns.regplot(x="Base MSRP",y="Electric Range",data=df,scatter={'alpha':0.5})
plt.title("ER vs Price")
plt.show()
```
